HTML로 데이터 가져오기

```python
import requests
from bs4 import BeautifulSoup

# 1. 웹페이지 주소
url = "https://quotes.toscrape.com/"

# 2. 웹페이지에 요청
response = requests.get(url, timeout=10)
# requests.get(url): 웹서버에 HTTP GET 요청을 보내고 응답을 받습니다.
# timeout=10 : 응답은 10초까지 기다림

response.raise_for_status()
# response.raise_for_status(): HTTP 오류 상태가 반환되었는지 확인합니다.

# 3. HTML 문자열 가져오기
html_text = response.text
# response.text: 응답 내용을 HTML 문자열로 가져옵니다.

# 4. HTML 구조 해석
soup = BeautifulSoup(html_text, "html.parser")
# BeautifulSoup(): HTML 문서의 구조를 해석합니다.


# 5. 첫 번째 명언 선택
quote = soup.select_one("div.quote")
# select_one(): CSS 선택자에 일치하는 첫 번째 HTML 요소를 찾습니다.

# 6. 명언 정보 확인
print("명언:", quote.select_one("span.text").get_text(strip=True))
print("작성자:", quote.select_one("small.author").get_text(strip=True))

tags = [
    tag.get_text(strip=True)    # get_text(): HTML 요소에 포함된 텍스트를 가져옵니다.
    for tag in quote.select("a.tag")
]

print("태그:", tags)
```

HTML과 JSON 데이터 추출하기

```python
import json
from bs4 import BeautifulSoup

# ----------------------
# 1. HTML 예제 데이터
# ----------------------

html_text = """
<div class="quote">
    <span class="text">첫 번째 예시 명언입니다.</span>
    <small class="author">작성자 A</small>
    <div class="tags">
        <a class="tag">life</a>
        <a class="tag">thinking</a>
    </div>
</div>

<div class="quote">
    <span class="text">두 번째 예시 명언입니다.</span>
    <small class="author">작성자 B</small>
    <div class="tags">
        <a class="tag">learning</a>
    </div>
</div>
"""

# ----------------------
# 2. HTML에서 데이터 추출
# ----------------------

soup = BeautifulSoup(html_text, "html.parser")

html_records = []

for quote in soup.select("div.quote"):
    text = quote.select_one("span.text").get_text(strip=True)

    author = quote.select_one(
        "small.author"
    ).get_text(strip=True)

    tags = [
        tag.get_text(strip=True)
        for tag in quote.select("a.tag")
    ]

    html_records.append({
        "text": text,
        "author": author,
        "tags": ", ".join(tags),
        "source_url": "https://quotes.toscrape.com/"
    })

# ----------------------
# 3. JSON 예제 데이터
# ----------------------

json_text = """
{
  "quotes": [
    {
      "text": "첫 번째 예시 명언입니다.",
      "author": {
        "name": "작성자 A"
      },
      "tags": ["life", "thinking"]
    },
    {
      "text": "두 번째 예시 명언입니다.",
      "author": {
        "name": "작성자 B"
      },
      "tags": ["learning"]
    }
  ]
}
"""

# ----------------------
# 4. JSON에서 데이터 추출
# ----------------------

data = json.loads(json_text)

json_records = []

for quote in data["quotes"]:
    json_records.append({
        "text": quote["text"],
        "author": quote["author"]["name"],
        "tags": ", ".join(quote["tags"]),
        "source_url": (
            "https://quotes.toscrape.com/api/quotes?page=1"
        )
    })

# ----------------------
# 5. 결과 확인
# ----------------------

print("HTML 추출 건수:", len(html_records))
print("JSON 추출 건수:", len(json_records))

print("\nHTML 추출 결과:")
for record in html_records:
    print(record)

print("\nJSON 추출 결과:")
for record in json_records:
    print(record)

# 두 형식에서 추출한 주요 필드 비교
fields = ["text", "author", "tags"]

html_values = [
    {field: row[field] for field in fields}
    for row in html_records
]

json_values = [
    {field: row[field] for field in fields}
    for row in json_records
]

print("\n주요 필드 일치 여부:", html_values == json_values)

# 실행 결과
#
# HTML 추출 건수: 2
# JSON 추출 건수: 2
#
# 주요 필드 일치 여부: True
```

실제 API 추출하기  
첫번째 페이지 get 요청 보내고, json응답에서 명언 데이터 추출

```python
import requests

# ----------------------
# 1. API 주소와 요청 조건
# ----------------------

url = "https://quotes.toscrape.com/api/quotes"

params = {
    "page": 1
}

# ----------------------
# 2. GET 요청과 JSON 파싱
# ----------------------

try:
    response = requests.get(
        url,
        params=params,
        timeout=10
    )

    response.raise_for_status()

    data = response.json()

except requests.exceptions.JSONDecodeError as error:
    print("JSON 파싱 실패:", error)

except requests.RequestException as error:
    print("API 요청 실패:", error)

else:
    # ----------------------
    # 3. 명언 목록 가져오기
    # ----------------------

    quotes = data["quotes"]

    # ----------------------
    # 4. 필요한 필드 추출
    # ----------------------

    records = []

    for quote in quotes:
        record = {
            "text": quote["text"],
            "author": quote["author"]["name"],
            "tags": ", ".join(quote["tags"]),
            "source_url": response.url
        }

        records.append(record)

    # ----------------------
    # 5. 수집 결과 확인
    # ----------------------

    print("현재 페이지:", data["page"])
    print("다음 페이지 존재:", data["has_next"])
    print("추출한 명언 수:", len(records))

    if records:
        print("추출한 필드:", list(records[0].keys()))
        print("첫 번째 작성자:", records[0]["author"])
        print("첫 번째 태그:", records[0]["tags"])
    else:
        print("추출된 명언이 없습니다.")
```

여러 페이지 수집하고 csv 저장하기

```python
import time

import pandas as pd
import requests

# ----------------------
# 1. API 및 수집 설정
# ----------------------

url = "https://quotes.toscrape.com/api/quotes"

records = []
page = 1

# 예상치 못한 무한 반복 방지
max_pages = 20

# ----------------------
# 2. 여러 페이지 수집
# ----------------------

try:
    while True:
        if page > max_pages:
            raise RuntimeError("페이지 수 상한을 초과했습니다.")

        # 현재 페이지 요청
        response = requests.get(
            url,
            params={"page": page},
            timeout=10
        )

        response.raise_for_status()

        # JSON 응답 파싱
        data = response.json()
        quotes = data["quotes"]

        if not isinstance(quotes, list) or not quotes:
            raise ValueError("명언 목록이 없거나 형식이 잘못되었습니다.")

        # ----------------------
        # 3. 필요한 필드 추출
        # ----------------------

        for quote in quotes:
            records.append({
                "text": quote["text"],
                "author": quote["author"]["name"],
                "tags": ", ".join(quote["tags"]),
                "source_url": response.url
            })

        print(
            f"페이지 {page}: "
            f"{len(quotes)}개 수집 / "
            f"누적 {len(records)}개"
        )

        # ----------------------
        # 4. 다음 페이지 확인
        # ----------------------

        if not data["has_next"]:
            break

        page += 1
        time.sleep(0.5)

    # ----------------------
    # 5. DataFrame 생성
    # ----------------------

    columns = [
        "text",
        "author",
        "tags",
        "source_url"
    ]

    df = pd.DataFrame(
        records,
        columns=columns
    )

    # ----------------------
    # 6. CSV 저장
    # ----------------------

    file_path = "quotes_raw.csv"

    df.to_csv(
        file_path,
        index=False,
        encoding="utf-8-sig"
    )

    # ----------------------
    # 7. 저장 결과 확인
    # ----------------------

    loaded_df = pd.read_csv(
        file_path,
        encoding="utf-8-sig"
    )

    print("\n전체 수집 건수:", len(df))
    print("저장 후 행 수:", len(loaded_df))
    print("컬럼:", loaded_df.columns.tolist())
    print("저장 파일:", file_path)

    print("\n데이터 미리보기:")
    print(loaded_df.head())

except (requests.RequestException, ValueError, KeyError,
        TypeError, RuntimeError) as error:
    print("데이터 수집 중단:", error)
    print("이번 실행의 CSV 저장을 완료하지 못했습니다.")
```

웹페이지에서 명언 추출하기

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

# ----------------------
# 1. 웹페이지 요청
# ----------------------

url = "https://quotes.toscrape.com/"

response = requests.get(
    url,
    timeout=10
)

response.raise_for_status()

# ----------------------
# 2. HTML 파싱
# ----------------------

soup = BeautifulSoup(
    response.text,
    "html.parser"
)

# ----------------------
# 3. 명언 영역 탐색
# ----------------------

quotes = soup.select("div.quote")

if not quotes:
    raise ValueError("HTML에서 명언 영역을 찾지 못했습니다.")

# ----------------------
# 4. 필요한 필드 추출
# ----------------------

records = []

for quote in quotes:
    text_element = quote.select_one("span.text")
    author_element = quote.select_one("small.author")
    tag_elements = quote.select("div.tags a.tag")

    # 필수 요소 확인
    if text_element is None or author_element is None:
        raise ValueError("명언 본문 또는 작성자 요소가 없습니다.")

    text = text_element.get_text(strip=True)
    author = author_element.get_text(strip=True)

    tags = [
        tag.get_text(strip=True)
        for tag in tag_elements
    ]

    records.append({
        "text": text,
        "author": author,
        "tags": ", ".join(tags),
        "source_url": response.url
    })

# ----------------------
# 5. DataFrame 생성
# ----------------------

columns = [
    "text",
    "author",
    "tags",
    "source_url"
]

df = pd.DataFrame(
    records,
    columns=columns
)

# ----------------------
# 6. 결과 확인
# ----------------------

print("HTTP 상태 코드:", response.status_code)
print("HTML 명언 영역 수:", len(quotes))
print("추출한 명언 수:", len(df))
print("DataFrame Shape:", df.shape)

print("\n첫 번째 명언의 작성자:")
print(df.loc[0, "author"])

print("\n데이터 미리보기:")
print(df.head())
```

중복, 결측, 수집 결과 확인하기

```python
from pathlib import Path
import pandas as pd

# ----------------------
# 1. CSV 파일 불러오기
# ----------------------

file_path = Path("quotes_raw.csv")

if not file_path.exists():
    raise FileNotFoundError(
        "quotes_raw.csv 파일이 없습니다."
    )

df = pd.read_csv(
    file_path,
    encoding="utf-8-sig",
    keep_default_na=False
)

# ----------------------
# 2. 필수 컬럼 확인
# ----------------------

expected_columns = [
    "text",
    "author",
    "tags",
    "source_url"
]

missing_columns = [
    col for col in expected_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"누락된 컬럼: {missing_columns}"
    )

print("전체 데이터 건수:", len(df))
print("컬럼:", df.columns.tolist())

# ----------------------
# 3. 결측값과 빈 문자열 확인
# ----------------------

# 원본을 유지하고 확인용 복사본 생성
check_df = df.copy()

for col in expected_columns:
    # 문자열 양쪽 공백 제거
    check_df[col] = (
        check_df[col]
        .astype("string")
        .str.strip()
    )

    # 빈 문자열을 결측값으로 표시
    check_df.loc[
        check_df[col] == "", col
    ] = pd.NA

print("\n컬럼별 결측·빈 문자열 건수:")
print(check_df[expected_columns].isna().sum())

# ----------------------
# 4. 중복 데이터 확인
# ----------------------

# 본문과 작성자가 모두 있는 행만 비교
valid_mask = (
    check_df["text"].notna()    # ".notna() 비어있지 않은 칸을 True로 표시
    & check_df["author"].notna()
)
# 값이 둘 다 있어야하면서 둘 다 같은지 확인

duplicate_mask = (
    valid_mask
    & check_df.duplicated(
        subset=["text", "author"],
        keep=False
    )
)

print("\n중복 후보 행 수:", duplicate_mask.sum())

print("\n중복 후보 데이터:")
print(
    check_df.loc[
        duplicate_mask,
        ["text", "author"]
    ].head(10)
)

# ----------------------
# 5. 수집 출처별 건수 확인
# ----------------------

print("\n수집 출처별 건수:")
print(
    check_df["source_url"].value_counts(
        dropna=False
    ).sort_index()
)

# ----------------------
# 6. 전체 확인 결과
# ----------------------

print("\n품질 탐색 결과")

print("전체 행 수:", len(check_df))

print(
    "본문 누락:",
    check_df["text"].isna().sum()
)

print(
    "작성자 누락:",
    check_df["author"].isna().sum()
)

print(
    "중복 후보:",
    duplicate_mask.sum()
)

print("\n원본 데이터는 변경하지 않았습니다.")
```

분석 가능한 데이터 건 수 확인하기  
- 중복과 결측을 실제로 제거하지 않고, 현재 데이터의 상태를 요약하는 코드

```python
import pandas as pd

# ----------------------
# 1. 데이터 불러오기
# ----------------------

df = pd.read_csv(
    "quotes_raw.csv",
    encoding="utf-8-sig",
    keep_default_na=False
)

# ----------------------
# 2. 필수 컬럼 확인
# ----------------------

required_columns = [
    "text",
    "author",
    "tags",
    "source_url"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"누락된 컬럼: {missing_columns}"
    )

# ----------------------
# 3. 공백을 제외한 값 존재 여부
# ----------------------

text = df["text"].fillna("").astype(str).str.strip()
author = df["author"].fillna("").astype(str).str.strip()
tags = df["tags"].fillna("").astype(str).str.strip()
source = df["source_url"].fillna("").astype(str).str.strip()

has_text = text.ne("") #.ne : = (이퀄이랑 같은 뜻)
has_author = author.ne("")
has_tags = tags.ne("")
has_source = source.ne("")

# ----------------------
# 4. 분석 목적별 활용 가능 건수
# ----------------------

text_count = has_text.sum()
author_count = (has_text & has_author).sum()
tag_count = (has_text & has_tags).sum()

# ----------------------
# 5. 중복 후보 확인
# ----------------------

duplicate_keys = pd.DataFrame({
    "text": text,
    "author": author
})

duplicate_mask = (
    has_text
    & has_author
    & duplicate_keys.duplicated(
        subset=["text", "author"],
        keep=False
    )
)

# ----------------------
# 6. 품질 확인 결과
# ----------------------

print("전체 수집 건수:", len(df))
print("본문이 있는 데이터:", text_count)
print("작성자별 분석 가능:", author_count)
print("태그별 분석 가능:", tag_count)
print("중복 후보 행 수:", duplicate_mask.sum())
print("출처 누락 건수:", (~has_source).sum())

print("\n분석 목적별 활용 가능 비율")

if len(df) > 0:
    print(
        "텍스트 분석:",
        f"{text_count / len(df):.1%}"
    )
    print(
        "작성자별 분석:",
        f"{author_count / len(df):.1%}"
    )
    print(
        "태그별 분석:",
        f"{tag_count / len(df):.1%}"
    )
else:
    print("수집된 데이터가 없습니다.")
```

데이터 정제하기

```python
from pathlib import Path
import pandas as pd

# ----------------------
# 1. 원시 데이터 불러오기
# ----------------------

file_path = Path("quotes_raw.csv")

if not file_path.exists():
    raise FileNotFoundError(
        "quotes_raw.csv 파일이 없습니다."
    )

raw_df = pd.read_csv(
    file_path,
    encoding="utf-8-sig",
    keep_default_na=False
)

# ----------------------
# 2. 필요한 컬럼 확인
# ----------------------

columns = [
    "text",
    "author",
    "tags",
    "source_url"
]

missing_columns = [
    col for col in columns
    if col not in raw_df.columns
]

if missing_columns:
    raise ValueError(
        f"누락된 컬럼: {missing_columns}"
    )

# 원본은 유지하고 정제용 복사본 생성
df = raw_df[columns].copy()

original_count = len(df)

# ----------------------
# 3. 문자열 공백 정리
# ----------------------

for col in columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
    )

    # 빈 문자열을 결측값으로 변환
    df[col] = df[col].replace("", pd.NA)

# ----------------------
# 4. 본문 누락 데이터 제외
# ----------------------

missing_text_count = df["text"].isna().sum()

df = df.dropna(
    subset=["text"]
).copy()

# ----------------------
# 5. 중복 데이터 제거
# ----------------------

before_dedup = len(df)

df = df.drop_duplicates(
    subset=["text", "author"],
    keep="first"
).reset_index(drop=True)

duplicate_removed = before_dedup - len(df)

# ----------------------
# 6. 정제 결과 확인
# ----------------------

print("원시 데이터 건수:", original_count)
print("본문 누락으로 제외:", missing_text_count)
print("중복으로 제외:", duplicate_removed)
print("정제 후 데이터 건수:", len(df))

print("\n남아 있는 결측값:")
print(df.isna().sum())

print("\n정제한 데이터 미리보기:")
print(df.head())

# ----------------------
# 7. 건수 검증
# ----------------------

assert len(df) == (
    original_count
    - missing_text_count
    - duplicate_removed
)

assert df["text"].notna().all()

assert not df.duplicated(
    subset=["text", "author"]
).any()

print("\n기본 정제 결과 확인 완료")
```

텍스트 분석용 데이터 셋 구성하기  
- quotes_raw.csv를 불러와 앞선 섹션의 정제 기준을 적용
- 분석용 DataFrame을 구성

```python
import pandas as pd
from pathlib import Path

# ----------------------
# 1. 원시 데이터 불러오기
# ----------------------

file_path = Path("quotes_raw.csv")

if not file_path.exists():
    raise FileNotFoundError(
        "quotes_raw.csv 파일이 없습니다."
    )

raw_df = pd.read_csv(
    file_path,
    encoding="utf-8-sig",
    keep_default_na=False
)

# ----------------------
# 2. 컬럼 구조 확인
# ----------------------

columns = [
    "text",
    "author",
    "tags",
    "source_url"
]

missing_columns = [
    col for col in columns
    if col not in raw_df.columns
]

if missing_columns:
    raise ValueError(
        f"누락된 컬럼: {missing_columns}"
    )

# ----------------------
# 3. 기본 정제 기준 적용
# ----------------------

df = raw_df[columns].copy()

# 문자열 형식과 공백 정리
for col in columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
    )

    df[col] = df[col].replace("", pd.NA)

# 본문이 없는 행 제외
df = df.dropna(
    subset=["text"]
)

# 동일 본문·작성자 중복 제거
df = df.drop_duplicates(
    subset=["text", "author"],
    keep="first"
)

# ----------------------
# 4. 최종 분석 컬럼 구성
# ----------------------

analysis_df = df[columns].copy()

# 선택적 정보의 결측값은 빈 문자열로 유지
optional_columns = [
    "author",
    "tags",
    "source_url"
]

analysis_df[optional_columns] = (
    analysis_df[optional_columns].fillna("")
)

analysis_df = analysis_df.reset_index(drop=True)

# ----------------------
# 5. 기본 구조 확인
# ----------------------

print("분석용 데이터 건수:", len(analysis_df))
print("Shape:", analysis_df.shape)
print("컬럼:", analysis_df.columns.tolist())

print("\n데이터 미리보기:")
print(analysis_df.head())

# ----------------------
# 6. 텍스트 분석 가능 여부 확인
# ----------------------

valid_text = (
    analysis_df["text"].notna()
    & analysis_df["text"].str.strip().ne("")
)

print("\n본문이 있는 데이터:", valid_text.sum())
print("작성자 없는 데이터:",
      analysis_df["author"].eq("").sum())
print("태그 없는 데이터:",
      analysis_df["tags"].eq("").sum())
print("출처 없는 데이터:",
      analysis_df["source_url"].eq("").sum())

assert valid_text.all()
assert analysis_df.columns.tolist() == columns

print("\n분석용 DataFrame 구성 완료")
```

최종 데이터셋 저장 및 검증

```python
from pathlib import Path
import pandas as pd

# ----------------------
# 1. 원시 데이터 불러오기
# ----------------------

raw_path = Path("quotes_raw.csv")
clean_path = Path("quotes_clean.csv")

if not raw_path.exists():
    raise FileNotFoundError(
        "quotes_raw.csv 파일이 없습니다."
    )

columns = [
    "text",
    "author",
    "tags",
    "source_url"
]

raw_df = pd.read_csv(
    raw_path,
    encoding="utf-8-sig",
    dtype="string",
    keep_default_na=False
)

# 필요한 컬럼 확인
missing_columns = [
    col for col in columns
    if col not in raw_df.columns
]

if missing_columns:
    raise ValueError(
        f"누락된 컬럼: {missing_columns}"
    )

# ----------------------
# 2. 기본 정제
# ----------------------

df = raw_df[columns].copy()

raw_count = len(df)

# 문자열 양쪽 공백 제거
for col in columns:
    df[col] = df[col].str.strip()
    df[col] = df[col].replace("", pd.NA)

# 본문이 없는 행 제외
missing_text_count = int(
    df["text"].isna().sum()
)

df = df.dropna(
    subset=["text"]
).copy()

# 중복 제거
before_dedup = len(df)

df = df.drop_duplicates(
    subset=["text", "author"],
    keep="first"
)

duplicate_removed = before_dedup - len(df)

df = df.reset_index(drop=True)

# ----------------------
# 3. 정제 결과 검증
# ----------------------

# 필수 본문 확인
assert df["text"].notna().all()

# 동일한 본문·작성자 중복 확인
assert not df.duplicated(
    subset=["text", "author"]
).any()

# 정제 전후 건수 확인
expected_count = (
    raw_count
    - missing_text_count
    - duplicate_removed
)

assert len(df) == expected_count

# 출처 누락 확인
missing_source_count = int(
    df["source_url"].isna().sum()
)

# ----------------------
# 4. 최종 CSV 저장
# ----------------------

# 선택적 항목의 결측값은 빈 문자열로 저장
clean_df = df.fillna("")

clean_df.to_csv(
    clean_path,
    index=False,
    encoding="utf-8-sig"
)

# ----------------------
# 5. 저장 파일 다시 불러오기
# ----------------------

saved_df = pd.read_csv(
    clean_path,
    encoding="utf-8-sig",
    dtype="string",
    keep_default_na=False
)

# ----------------------
# 6. 최종 검증
# ----------------------

# 컬럼 구조 확인
assert saved_df.columns.tolist() == columns

# 데이터 건수 확인
assert len(saved_df) == len(clean_df)

# 저장 전후 데이터 내용 확인
assert clean_df.equals(saved_df)

print("원시 데이터:", raw_count)
print("본문 누락 제거:", missing_text_count)
print("중복 제거:", duplicate_removed)
print("최종 데이터:", len(saved_df))

print("\n출처 누락:", missing_source_count)

print("\n최종 컬럼:")
print(saved_df.columns.tolist())

print("\n최종 데이터 미리보기:")
print(saved_df.head())

print("\n저장 파일:", clean_path)

if missing_source_count == 0:
    print("기본 데이터 품질 검증 완료")
else:
    print("출처 누락 데이터에 대한 추가 확인 필요")
```
